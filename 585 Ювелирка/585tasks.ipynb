Рассчитайте наценку на ювелирные изделия по данным на листе "Задание 2". Соблюсти следующие условия:
1. Период анализа - с 1.04.23 по 10.04.23
2. В анализе должны участвовать только продажи изделий с общей суммой и себестоимостью больше 0 рублей
3. В ответе необходимо получить два значения наценки - среднюю наценку на ювелирные изделия из золота и среднюю наценку на ювелирные изделия из серебра. Сгруппировать изделия по металлу нужно на основе поля "Товарная группа". Товарная группа изделий из серебра начинается на "СИ", все остальные изделия - это золото.

Для начала загрузим данные в pandas и выберем нужную страницу для данных.

```python
import pandas as pd

file_path = "Т•бвЃҐЃ• 3.xlsx"

excel_file = pd.ExcelFile(file_path)
print("Names of lists", excel_file.sheet_names)
```

```python
df = pd.read_excel(
    file_path,
    sheet_name="Задание 2"
)
df.head()
```

```python
df.dtypes
```

```python
# 1. Период анализа - с 1.04.23 по 10.04.23
filtered_df = df[df["ДатаПродажи"].between("2023-04-01",
                                 "2023-04-11",
                                 inclusive="left")]
# 2. В анализе должны участвовать только продажи изделий с общей суммой и себестоимостью больше 0 рублей
filtered_df = filtered_df[filtered_df["ОбщаяСумма"] > 0]
filtered_df = filtered_df[filtered_df["Себестоимость"] > 0]
```

Первые 2 пункта сделали, делаем третий:

3. В ответе необходимо получить два значения наценки - среднюю наценку на ювелирные изделия из золота и среднюю наценку на ювелирные изделия из серебра. Сгруппировать изделия по металлу нужно на основе поля "Товарная группа". Товарная группа изделий из серебра начинается на "СИ", все остальные изделия - это золото.

```python
mask_silver = filtered_df["ТоварнаяГруппа"].str.startswith("СИ")
silver_df = filtered_df[mask_silver]
gold_df = filtered_df[~mask_silver]

silver_df["Наценка"] = (silver_df["ОбщаяСумма"] - silver_df["Себестоимость"]) / silver_df["Себестоимость"] * 100
gold_df["Наценка"] = (gold_df["ОбщаяСумма"] - gold_df["Себестоимость"]) / gold_df["Себестоимость"] * 100

print(f"Средняя наценка на товары из серебра: {silver_df["Наценка"].mean():.3f}%")
print(f"Средняя наценка на товары из золота: {gold_df["Наценка"].mean():.3f}%")
```
